```python
# 3. Example 1: Disaster Ground Truth Context (Tavily Primary)
req_disaster = WebIntelligenceRequest(
    query="July 2023 Delhi Yamuna flood Hathnikund barrage discharge water level",
    max_results=3,
    location_hint="Delhi, India"
)
res_disaster = fetch_web_intelligence(req_disaster)
print(f"Provider used: {res_disaster.get('provider')}")
print(f"Summary: {res_disaster.get('summary')}\n")
for r in res_disaster.get('results', []):
    print(f"- {r['title']} ({r['url']})")
```
